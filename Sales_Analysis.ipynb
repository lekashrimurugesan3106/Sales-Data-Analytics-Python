```python
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error
from sklearn.linear_model import LinearRegression
import numpy as np

# Data setup
dates = pd.date_range(start='2023-01-01', periods=36, freq='M')
df_val = pd.DataFrame({
    'Date': dates,
    'Maths': 70 + np.arange(36)*0.8 + 5*np.sin(np.arange(36)*0.5) + np.random.normal(0,2,36)
})
df_val['Month_Num'] = range(len(df_val))

# TimeSeriesSplit - future data train la vara kudathu
tscv = TimeSeriesSplit(n_splits=5)
X = df_val[['Month_Num']]
y = df_val['Maths']

mae_scores = []
fold = 1

for train_idx, test_idx in tscv.split(X):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = LinearRegression()
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    mae = mean_absolute_error(y_test, preds)
    mae_scores.append(mae)
    print(f"Fold {fold} MAE: {mae:.2f}")
    fold += 1

print(f"\nAverage MAE: {np.mean(mae_scores):.2f}")

# Plot splits
plt.figure(figsize=(12,5))
for i, (train_idx, test_idx) in enumerate(tscv.split(X)):
    plt.plot(train_idx, [i]*len(train_idx), 'b.')
    plt.plot(test_idx, [i]*len(test_idx), 'r.')
plt.title('TimeSeriesSplit - Blue=Train, Red=Test')
plt.xlabel('Data Index')
plt.ylabel('Fold')
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

# Sample data - trend irukura data
dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
df_stat = pd.DataFrame({
    'Date': dates,
    'Maths': 70 + np.arange(100)*0.2 + np.random.normal(0, 2, 100)
})

# ADF Test - Stationary ah nu check
def adf_test(series):
    result = adfuller(series.dropna())
    print(f'ADF Statistic: {result[0]:.3f}')
    print(f'p-value: {result[1]:.3f}')
    if result[1] <= 0.05:
        print("Stationary: ARIMA use pannalam")
    else:
        print("Non-Stationary: Differencing pannanum")

print("Original Data:")
adf_test(df_stat['Maths'])

# Differencing - trend remove panna
df_stat['Maths_diff'] = df_stat['Maths'].diff()

print("\nAfter Differencing:")
adf_test(df_stat['Maths_diff'])

# Plot
fig, axes = plt.subplots(2,1, figsize=(12,6))
axes[0].plot(df_stat['Date'], df_stat['Maths'])
axes[0].set_title('Original - Trend iruku, Non-Stationary')
axes[1].plot(df_stat['Date'], df_stat['Maths_diff'])
axes[1].set_title('After Diff - Stationary')
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# Sample data
dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
np.random.seed(0)
df_walk = pd.DataFrame({
    'Date': dates,
    'Maths': 75 + np.arange(100)*0.05 + 3*np.sin(np.arange(100)*0.2) + np.random.normal(0,1,100)
})

# Lag feature
df_walk['Maths_Lag1'] = df_walk['Maths'].shift(1)
df_walk = df_walk.dropna()

# Walk Forward Setup
n_train = 60   # First 60 days train
n_test = 20    # 20 days ahead test
step = 1       # 1 day ah slide pannu

predictions = []
actuals = []
dates_pred = []

for i in range(n_train, len(df_walk) - n_test, step):
    train = df_walk.iloc[i-n_train:i]
    test = df_walk.iloc[i:i+n_test]

    X_train = train[['Maths_Lag1']]
    y_train = train['Maths']
    X_test = test[['Maths_Lag1']]
    y_test = test['Maths']

    model = LinearRegression()
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    predictions.extend(pred)
    actuals.extend(y_test)
    dates_pred.extend(test['Date'])

mae = mean_absolute_error(actuals, predictions)
print(f"Walk Forward MAE: {mae:.2f}")

# Plot
plt.figure(figsize=(14,6))
plt.plot(df_walk['Date'], df_walk['Maths'], label='Actual Full Data', alpha=0.5)
plt.plot(dates_pred, predictions, 'r.', label='Walk Forward Predictions')
plt.title('Walk Forward Validation - Production Style Testing')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler

# Sample data
dates = pd.date_range(start='2023-01-01', periods=200, freq='D')
data = 75 + np.arange(200)*0.05 + 10*np.sin(np.arange(200)*0.1) + np.random.normal(0,1,200)
df_lstm = pd.DataFrame({'Date': dates, 'Maths': data})

# Scale panna - LSTM ku 0-1 range venum
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df_lstm[['Maths']])

# Sequence create pannu - 10 days vachu next day predict
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

seq_len = 10
X, y = create_sequences(scaled_data, seq_len)

# Train-Test split
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# LSTM Model
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(50, activation='relu', input_shape=(seq_len, 1)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=50, batch_size=16, verbose=0)

# Predict
pred_scaled = model.predict(X_test)
predictions = scaler.inverse_transform(pred_scaled)
actual = scaler.inverse_transform(y_test)

mae = mean_absolute_error(actual, predictions)
print(f"LSTM MAE: {mae:.2f}")

# Plot
plt.figure(figsize=(14,6))
plt.plot(df_lstm['Date'][seq_len+split:], actual, label='Actual')
plt.plot(df_lstm['Date'][seq_len+split:], predictions, label='LSTM Predicted')
plt.title('LSTM Time Series Forecast')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
from scipy import stats

# Sample data - anomalies inject pannirukom
dates = pd.date_range(start='2024-01-01', periods=100, freq='D')
np.random.seed(1)
data = 80 + np.random.normal(0, 3, 100)
data[20] = 120 # Anomaly 1
data[50] = 40 # Anomaly 2
data[85] = 115 # Anomaly 3

df_anom = pd.DataFrame({'Date': dates, 'Maths': data})

# Method 1: Z-Score - threshold > 3 na outlier
df_anom['Z_Score'] = np.abs(stats.zscore(df_anom['Maths']))
df_anom['Z_Anomaly'] = df_anom['Z_Score'] > 3

# Method 2: Rolling Mean + Std - sudden spike pidikka
window = 7
df_anom['Roll_Mean'] = df_anom['Maths'].rolling(window).mean()
df_anom['Roll_Std'] = df_anom['Maths'].rolling(window).std()
df_anom['Upper'] = df_anom['Roll_Mean'] + 2.5*df_anom['Roll_Std']
df_anom['Lower'] = df_anom['Roll_Mean'] - 2.5*df_anom['Roll_Std']
df_anom['Roll_Anomaly'] = (df_anom['Maths'] > df_anom['Upper']) | (df_anom['Maths'] < df_anom['Lower'])

# Method 3: Isolation Forest - ML based
iso = IsolationForest(contamination=0.03, random_state=42)
df_anom['IF_Anomaly'] = iso.fit_predict(df_anom[['Maths']]) == -1

# Plot
plt.figure(figsize=(14,7))
plt.plot(df_anom['Date'], df_anom['Maths'], label='Maths', marker='o', alpha=0.6)
plt.plot(df_anom['Date'], df_anom['Roll_Mean'], 'g--', label='Rolling Mean')
plt.fill_between(df_anom['Date'], df_anom['Lower'], df_anom['Upper'], color='gray', alpha=0.2, label='2.5 Std Band')

# Anomalies mark pannu
anoms = df_anom[df_anom['IF_Anomaly']]
plt.scatter(anoms['Date'], anoms['Maths'], color='red', s=100, label='Anomaly', zorder=5)

plt.title('Anomaly Detection - Isolation Forest + Rolling Stats')
plt.legend()
plt.grid(True)
plt.show()

print("Detected Anomalies:")
print(df_anom[df_anom['IF_Anomaly']][['Date','Maths']])
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller

# Sample data - trend + seasonality iruku
dates = pd.date_range(start='2022-01-01', periods=36, freq='M')
np.random.seed(4)
trend = np.linspace(70, 85, 36)
season = 5*np.sin(2*np.pi*np.arange(36)/12)
noise = np.random.normal(0, 1, 36)
data = trend + season + noise

df_arima = pd.DataFrame({'Date': dates, 'Maths': data})
df_arima = df_arima.set_index('Date')

# Stationarity check
adf_result = adfuller(df_arima['Maths'])
print(f"ADF p-value: {adf_result[1]:.4f}") # <0.05 na stationary

# Diff pannu - d=1
df_arima['Maths_diff'] = df_arima['Maths'].diff().dropna()

# SARIMA model - (p,d,q)(P,D,Q,s)
# p,d,q = non-seasonal | P,D,Q,s = seasonal | s=12 monthly data
model = SARIMAX(
    df_arima['Maths'],
    order=(1,1,1), # p,d,q
    seasonal_order=(1,1,0,12), # P,D,Q,s
    enforce_stationarity=False,
    enforce_invertibility=False
)
results = model.fit(disp=False)

# Forecast - next 12 months
forecast = results.get_forecast(steps=12)
pred_mean = forecast.predicted_mean
pred_ci = forecast.conf_int()

print("\nSARIMA Summary:")
print(results.summary().tables[1]) # Coefficients

# Plot
plt.figure(figsize=(14,6))
plt.plot(df_arima.index, df_arima['Maths'], label='Actual', marker='o')
plt.plot(pred_mean.index, pred_mean, label='SARIMA Forecast', color='red')
plt.fill_between(pred_ci.index, pred_ci.iloc[:,0], pred_ci.iloc[:,1], color='pink', alpha=0.3)
plt.title('SARIMA Forecast - Next 12 Months')
plt.legend()
plt.grid(True)
plt.show()

# Residual check
residuals = results.resid
print(f"\nResiduals Mean: {residuals.mean():.4f}") # 0 kitta irukanum
```

```python
*Time Series Ensembling - Multiple models combine panna*

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error

# Sample data
dates = pd.date_range(start='2022-01-01', periods=48, freq='M')
np.random.seed(5)
trend = np.linspace(70, 90, 48)
season = 5*np.sin(2*np.pi*np.arange(48)/12)
noise = np.random.normal(0, 1.5, 48)
data = trend + season + noise

df_ens = pd.DataFrame({'Date': dates, 'Maths': data})
df_ens = df_ens.set_index('Date')

# Train-Test split
train = df_ens.iloc[:-6]
test = df_ens.iloc[-6:]

# Model 1: Linear Regression with lags
train_lr = train.copy()
train_lr['Lag1'] = train_lr['Maths'].shift(1)
train_lr['Lag12'] = train_lr['Maths'].shift(12)
train_lr = train_lr.dropna()

lr = LinearRegression()
lr.fit(train_lr[['Lag1','Lag12']], train_lr['Maths'])

test_lr = test.copy()
test_lr['Lag1'] = df_ens['Maths'].shift(1)[-6:]
test_lr['Lag12'] = df_ens['Maths'].shift(12)[-6:]
pred_lr = lr.predict(test_lr[['Lag1','Lag12']])

# Model 2: Random Forest
train_rf = train_lr.copy()
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(train_rf[['Lag1','Lag12']], train_rf['Maths'])
pred_rf = rf.predict(test_lr[['Lag1','Lag12']])

# Model 3: SARIMA
sarima = SARIMAX(train['Maths'], order=(1,1,1), seasonal_order=(1,0,0,12))
sarima_fit = sarima.fit(disp=False)
pred_sarima = sarima_fit.forecast(steps=6)

# Ensemble - Simple Average
pred_ensemble = (pred_lr + pred_rf + pred_sarima) / 3

# Weighted Average - SARIMA ku adhiga weight
pred_weighted = 0.2*pred_lr + 0.3*pred_rf + 0.5*pred_sarima

# Metrics
print("MAE LinearReg:", mean_absolute_error(test['Maths'], pred_lr))
print("MAE RandomForest:", mean_absolute_error(test['Maths'], pred_rf))
print("MAE SARIMA:", mean_absolute_error(test['Maths'], pred_sarima))
print("MAE Ensemble Avg:", mean_absolute_error(test['Maths'], pred_ensemble))
print("MAE Weighted Avg:", mean_absolute_error(test['Maths'], pred_weighted))

# Plot
plt.figure(figsize=(14,6))
plt.plot(df_ens.index, df_ens['Maths'], 'k-', label='Actual')
plt.plot(test.index, pred_lr, 'b--', label='LinearReg')
plt.plot(test.index, pred_rf, 'g--', label='RandomForest')
plt.plot(test.index, pred_sarima, 'r--', label='SARIMA')
plt.plot(test.index, pred_ensemble, 'm-', linewidth=2, label='Ensemble Avg')
plt.title('Time Series Ensembling - 3 Models Combined')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
import matplotlib.pyplot as plt

# Sample data
dates = pd.date_range(start='2022-01-01', periods=200, freq='D')
np.random.seed(14)
data = 60 + 0.2*np.arange(200) + 8*np.sin(np.arange(200)*0.3) + np.random.normal(0,1.5,200)
df = pd.DataFrame({'Date': dates, 'Value': data})

# Normalize data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[['Value']])

# Create sequences
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

seq_length = 14
X, y = create_sequences(scaled_data, seq_length)

# Train-Test split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# Build LSTM model
model = Sequential()
model.add(LSTM(50, activation='relu', input_shape=(seq_length, 1)))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mse')

# Train
model.fit(X_train, y_train, epochs=50, batch_size=16, verbose=0)

# Predict
y_pred = model.predict(X_test, verbose=0)

# Inverse transform
y_test_inv = scaler.inverse_transform(y_test)
y_pred_inv = scaler.inverse_transform(y_pred)

# Evaluate
mae = mean_absolute_error(y_test_inv, y_pred_inv)
print(f"LSTM MAE: {mae:.3f}")

# Plot
plt.figure(figsize=(15,6))
test_dates = df['Date'].iloc[train_size+seq_length:]
plt.plot(test_dates, y_test_inv, label='Actual')
plt.plot(test_dates, y_pred_inv, label='LSTM Forecast', linestyle='--')
plt.title('LSTM Time Series Forecast')
plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error
import matplotlib.pyplot as plt

# Sample data
dates = pd.date_range(start='2022-01-01', periods=250, freq='D')
np.random.seed(23)
trend = 85 + 0.08*np.arange(250)
seasonal = 12*np.sin(2*np.pi*np.arange(250)/14) + 7*np.cos(2*np.pi*np.arange(250)/30)
noise = np.random.normal(0, 2.5, 250)
data = trend + seasonal + noise
df = pd.DataFrame({'Date': dates, 'Value': data})

# Scale data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[['Value']])

# Create sequences
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

seq_length = 20
X, y = create_sequences(scaled_data, seq_length)

# Train-Test split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# Convert to torch tensors
X_train = torch.FloatTensor(X_train)
y_train = torch.FloatTensor(y_train)
X_test = torch.FloatTensor(X_test)
y_test = torch.FloatTensor(y_test)

# LSTM Model
class LSTMModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2):
        super(LSTMModel, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)
        out, _ = self.lstm(x, (h0, c0))
        out = self.fc(out[:, -1, :])
        return out

model = LSTMModel()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# Train
epochs = 100
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    outputs = model(X_train)
    loss = criterion(outputs, y_train)
    loss.backward()
    optimizer.step()
    if (epoch+1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.6f}')

# Predict
model.eval()
with torch.no_grad():
    train_pred = model(X_train)
    test_pred = model(X_test)

# Inverse transform
train_pred = scaler.inverse_transform(train_pred.numpy())
y_train_inv = scaler.inverse_transform(y_train.numpy())
test_pred = scaler.inverse_transform(test_pred.numpy())
y_test_inv = scaler.inverse_transform(y_test.numpy())

# Evaluate
mae = mean_absolute_error(y_test_inv, test_pred)
print(f"\nLSTM MAE: {mae:.3f}")

# Plot
train_plot = np.full(len(df), np.nan)
train_plot[seq_length:seq_length+len(train_pred)] = train_pred.flatten()

test_plot = np.full(len(df), np.nan)
test_plot[seq_length+len(train_pred):seq_length+len(train_pred)+len(test_pred)] = test_pred.flatten()

plt.figure(figsize=(15,6))
plt.plot(df['Date'], df['Value'], label='Actual')
plt.plot(df['Date'], train_plot, label='Train Pred', color='orange')
plt.plot(df['Date'], test_plot, label='Test Pred', color='red')
plt.title('LSTM Time Series Forecast')
plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()

# Next 7 days forecast
last_seq = scaled_data[-seq_length:]
future_pred = []
model.eval()
with torch.no_grad():
    for _ in range(7):
        seq_tensor = torch.FloatTensor(last_seq).unsqueeze(0)
        pred = model(seq_tensor).item()
        future_pred.append(pred)
        last_seq = np.append(last_seq[1:], [[pred]], axis=0)

future_pred = scaler.inverse_transform(np.array(future_pred).reshape(-1,1))
future_dates = pd.date_range(start=df['Date'].iloc[-1] + pd.Timedelta(days=1), periods=7)

print("\nForecast for next 7 days:")
for date, val in zip(future_dates, future_pred):
    print(f"{date.date()}: {val[0]:.2f}")
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error
import matplotlib.pyplot as plt

# Sample data
dates = pd.date_range(start='2022-01-01', periods=300, freq='D')
np.random.seed(25)
trend = 90 + 0.1*np.arange(300)
seasonal = 18*np.sin(2*np.pi*np.arange(300)/7) + 10*np.sin(2*np.pi*np.arange(300)/90)
noise = np.random.normal(0, 2.5, 300)
data = trend + seasonal + noise
df = pd.DataFrame({'Date': dates, 'Value': data})

# Scale data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[['Value']])

# Create sequences
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

seq_length = 60 # TCN needs longer context
X, y = create_sequences(scaled_data, seq_length)

# Train-Test split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# Convert to torch: TCN expects [batch, channels, seq_len]
X_train = torch.FloatTensor(X_train).transpose(1, 2)
y_train = torch.FloatTensor(y_train)
X_test = torch.FloatTensor(X_test).transpose(1, 2)
y_test = torch.FloatTensor(y_test)

# TCN Block
class Chomp1d(nn.Module):
    def __init__(self, chomp_size):
        super(Chomp1d, self).__init__()
        self.chomp_size = chomp_size

    def forward(self, x):
        return x[:, :, :-self.chomp_size].contiguous()

class TemporalBlock(nn.Module):
    def __init__(self, n_inputs, n_outputs, kernel_size, stride, dilation, padding, dropout=0.2):
        super(TemporalBlock, self).__init__()
        self.conv1 = nn.Conv1d(n_inputs, n_outputs, kernel_size,
                               stride=stride, padding=padding, dilation=dilation)
        self.chomp1 = Chomp1d(padding)
        self.relu1 = nn.ReLU()
        self.dropout1 = nn.Dropout(dropout)

        self.conv2 = nn.Conv1d(n_outputs, n_outputs, kernel_size,
                               stride=stride, padding=padding, dilation=dilation)
        self.chomp2 = Chomp1d(padding)
        self.relu2 = nn.ReLU()
        self.dropout2 = nn.Dropout(dropout)

        self.net = nn.Sequential(self.conv1, self.chomp1, self.relu1, self.dropout1,
                                 self.conv2, self.chomp2, self.relu2, self.dropout2)
        self.downsample = nn.Conv1d(n_inputs, n_outputs, 1) if n_inputs!= n_outputs else None
        self.relu = nn.ReLU()

    def forward(self, x):
        out = self.net(x)
        res = x if self.downsample is None else self.downsample(x)
        return self.relu(out + res)

class TCN(nn.Module):
    def __init__(self, input_size, output_size, num_channels, kernel_size=2, dropout=0.2):
        super(TCN, self).__init__()
        layers = []
        num_levels = len(num_channels)
        for i in range(num_levels):
            dilation_size = 2 ** i
            in_channels = input_size if i == 0 else num_channels[i-1]
            out_channels = num_channels[i]
            layers += [TemporalBlock(in_channels, out_channels, kernel_size, stride=1, dilation=dilation_size,
                                     padding=(kernel_size-1) * dilation_size, dropout=dropout)]

        self.network = nn.Sequential(*layers)
        self.linear = nn.Linear(num_channels[-1], output_size)

    def forward(self, x):
        out = self.network(x)
        out = out[:, :, -1] # Take last time step
        out = self.linear(out)
        return out

# Model: input_size=1, output=1, channels=[64, 64, 64, 64]
model = TCN(input_size=1, output_size=1, num_channels=[64, 64, 64, 64], kernel_size=3, dropout=0.2)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.002)

# Train
epochs = 120
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    outputs = model(X_train)
    loss = criterion(outputs, y_train)
    loss.backward()
    optimizer.step()
    if (epoch+1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.6f}')

# Predict
model.eval()
with torch.no_grad():
    train_pred = model(X_train)
    test_pred = model(X_test)

# Inverse transform
train_pred = scaler.inverse_transform(train_pred.numpy())
y_train_inv = scaler.inverse_transform(y_train.numpy())
test_pred = scaler.inverse_transform(test_pred.numpy())
y_test_inv = scaler.inverse_transform(y_test.numpy())

# Evaluate
mae = mean_absolute_error(y_test_inv, test_pred)
print(f"\nTCN MAE: {mae:.3f}")

# Plot
train_plot = np.full(len(df), np.nan)
train_plot[seq_length:seq_length+len(train_pred)] = train_pred.flatten()

test_plot = np.full(len(df), np.nan)
test_plot[seq_length+len(train_pred):seq_length+len(train_pred)+len(test_pred)] = test_pred.flatten()

plt.figure(figsize=(15,6))
plt.plot(df['Date'], df['Value'], label='Actual')
plt.plot(df['Date'], train_plot, label='Train Pred', color='orange')
plt.plot(df['Date'], test_plot, label='Test Pred', color='purple')
plt.title('TCN - Temporal Convolutional Network Forecast')
plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()

# Next 21 days forecast
last_seq = scaled_data[-seq_length:].reshape(1, -1) # [1, seq_len]
future_pred = []
model.eval()
with torch.no_grad():
    for _ in range(21):
        seq_tensor = torch.FloatTensor(last_seq).unsqueeze(1) # [1, 1, seq_len]
        pred = model(seq_tensor).item()
        future_pred.append(pred)
        last_seq = np.hstack([last_seq[:, 1:], [[pred]]])

future_pred = scaler.inverse_transform(np.array(future_pred).reshape(-1,1))
future_dates = pd.date_range(start=df['Date'].iloc[-1] + pd.Timedelta(days=1), periods=21)

print("\nTCN Forecast for next 21 days:")
for date, val in zip(future_dates, future_pred):
    print(f"{date.date()}: {val[0]:.2f}")
```

```python
from darts import TimeSeries
from darts.models import TransformerModel
from darts.metrics import mae
from darts.dataprocessing.transformers import Scaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample data
dates = pd.date_range(start='2022-01-01', periods=350, freq='D')
np.random.seed(26)
trend = 100 + 0.15*np.arange(350)
seasonal = 20*np.sin(2*np.pi*np.arange(350)/7) + 12*np.sin(2*np.pi*np.arange(350)/30) + 8*np.sin(2*np.pi*np.arange(350)/365)
noise = np.random.normal(0, 3, 350)
data = trend + seasonal + noise
df = pd.DataFrame({'Date': dates, 'Value': data})

# Convert to Darts TimeSeries
series = TimeSeries.from_dataframe(df, 'Date', 'Value')

# Train/Val split
train, val = series.split_before(0.8)

# Scale
scaler = Scaler()
train_scaled = scaler.fit_transform(train)
val_scaled = scaler.transform(val)
series_scaled = scaler.transform(series)

# Transformer Model - Attention based
model = TransformerModel(
    input_chunk_length=60, # past 60 days use panni predict
    output_chunk_length=14, # next 14 days predict
    batch_size=32,
    n_epochs=200,
    model_name="transformer_ts",
    nr_epochs_val_period=10,
    d_model=64,
    nhead=8,
    num_encoder_layers=3,
    num_decoder_layers=3,
    dim_feedforward=256,
    dropout=0.1,
    optimizer_kwargs={"lr": 1e-3},
    random_state=42,
    force_reset=True,
    save_checkpoints=True,
)

# Train
model.fit(train_scaled, val_series=val_scaled, verbose=True)

# Predict on validation
pred = model.predict(n=len(val))
pred = scaler.inverse_transform(pred)

# Evaluate
val_actual = scaler.inverse_transform(val_scaled)
mae_val = mae(val_actual, pred)
print(f"\nTransformer MAE: {mae_val:.3f}")

# Plot
plt.figure(figsize=(15,6))
series.plot(label='Actual')
pred.plot(label='Transformer Forecast', lw=2)
plt.title('Transformer Model Forecast')
plt.legend()
plt.grid(True)
plt.show()

# Future forecast 30 days
future_forecast = model.predict(n=30)
future_forecast = scaler.inverse_transform(future_forecast)

print("\nTransformer Forecast for next 30 days:")
for i in range(30):
    date = future_forecast.time_index[i]
    value = future_forecast.values()[i][0]
    print(f"{date.date()}: {value:.2f}")

# Backtest - rolling forecast
backtest = model.historical_forecasts(
    series_scaled,
    start=0.7,
    forecast_horizon=14,
    stride=7,
    retrain=False,
    verbose=True
)
backtest = scaler.inverse_transform(backtest)

plt.figure(figsize=(15,6))
series.plot(label='Actual')
backtest.plot(label='Backtest 14-day forecasts')
plt.title('Transformer Rolling Backtest')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from prophet import Prophet
import xgboost as xgb
from sklearn.metrics import mean_absolute_error
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data
dates = pd.date_range(start='2021-01-01', periods=600, freq='D')
np.random.seed(28)
trend = 150 + 0.25*np.arange(600)
weekly = 30*np.sin(2*np.pi*np.arange(600)/7)
monthly = 20*np.sin(2*np.pi*np.arange(600)/30)
noise = np.random.normal(0, 5, 600)
data = trend + weekly + monthly + noise
df = pd.DataFrame({'ds': dates, 'y': data})

# Prophet
m = Prophet(yearly_seasonality=True, weekly_seasonality=True, daily_seasonality=False)
m.fit(df.iloc[:-60])
future = m.make_future_dataframe(periods=60)
forecast = m.predict(future)

# XGBoost on Prophet residuals
train_residuals = df.iloc[:-60]['y'].values - forecast.iloc[:-60]['yhat'].values

def create_features(data, lag=14):
    X, y = [], []
    for i in range(lag, len(data)):
        X.append(data[i-lag:i])
        y.append(data[i])
    return np.array(X), np.array(y)

X_train, y_train = create_features(train_residuals, lag=14)

xgb_model = xgb.XGBRegressor(n_estimators=200, max_depth=5, learning_rate=0.05)
xgb_model.fit(X_train, y_train)

# Forecast residuals
last_resid = train_residuals[-14:]
resid_pred = []
for _ in range(60):
    pred = xgb_model.predict(last_resid.reshape(1,-1))[0]
    resid_pred.append(pred)
    last_resid = np.append(last_resid[1:], pred)

# Final hybrid forecast
hybrid_forecast = forecast.iloc[-60:]['yhat'].values + resid_pred

# Evaluate
mae_prophet = mean_absolute_error(df.iloc[-60:]['y'], forecast.iloc[-60:]['yhat'])
mae_hybrid = mean_absolute_error(df.iloc[-60:]['y'], hybrid_forecast)
print(f"Prophet MAE: {mae_prophet:.3f}")
print(f"Hybrid MAE: {mae_hybrid:.3f}")

# Plot
plt.figure(figsize=(15,6))
plt.plot(df['ds'], df['y'], label='Actual')
plt.plot(forecast['ds'], forecast['yhat'], label='Prophet')
plt.plot(forecast.iloc[-60:]['ds'], hybrid_forecast, label='Prophet+XGBoost')
plt.legend()
plt.grid(True)
plt.show()

print("\nHybrid Forecast for next 30 days:")
for i in range(30):
    date = forecast.iloc[-60+i]['ds'].date()
    print(f"{date}: {hybrid_forecast[i]:.2f}")
```

```python
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data
dates = pd.date_range(start='2020-01-01', periods=800, freq='D')
np.random.seed(29)
trend = 200 + 0.3*np.arange(800)
weekly = 35*np.sin(2*np.pi*np.arange(800)/7)
monthly = 25*np.sin(2*np.pi*np.arange(800)/30)
yearly = 15*np.sin(2*np.pi*np.arange(800)/365.25)
noise = np.random.normal(0, 6, 800)
data = trend + weekly + monthly + yearly + noise
df = pd.DataFrame({'Date': dates, 'Value': data})

# Feature Engineering
def create_features(df, label=None):
    df = df.copy()
    df['dayofweek'] = df['Date'].dt.dayofweek
    df['month'] = df['Date'].dt.month
    df['day'] = df['Date'].dt.day
    df['dayofyear'] = df['Date'].dt.dayofyear
    df['quarter'] = df['Date'].dt.quarter
    df['year'] = df['Date'].dt.year
    df['weekofyear'] = df['Date'].dt.isocalendar().week

    for lag in [1,7,14,28,30,60]:
        df[f'lag_{lag}'] = df['Value'].shift(lag)

    for window in [7,14,28]:
        df[f'rolling_mean_{window}'] = df['Value'].shift(1).rolling(window).mean()
        df[f'rolling_std_{window}'] = df['Value'].shift(1).rolling(window).std()

    df = df.dropna()
    X = df[['dayofweek','month','day','dayofyear','quarter','year','weekofyear',
            'lag_1','lag_7','lag_14','lag_28','lag_30','lag_60',
            'rolling_mean_7','rolling_mean_14','rolling_mean_28',
            'rolling_std_7','rolling_std_14','rolling_std_28']]
    if label:
        y = df[label]
        return X, y
    return X

# Train/Test split
train_size = int(len(df) * 0.8)
train_df = df.iloc[:train_size].copy()
test_df = df.iloc[train_size:].copy()

X_train, y_train = create_features(train_df, 'Value')
X_test, y_test = create_features(pd.concat([train_df.iloc[-60:], test_df]), 'Value')
X_test = X_test.iloc[-len(test_df):]

# LightGBM
model = lgb.LGBMRegressor(
    n_estimators=1000,
    learning_rate=0.01,
    max_depth=5,
    num_leaves=31,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

model.fit(X_train, y_train)

# Predict
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, test_pred)
print(f"LightGBM MAE: {mae:.3f}")

# Plot
plt.figure(figsize=(15,6))
plt.plot(train_df['Date'], y_train, label='Train Actual')
plt.plot(test_df['Date'], y_test, label='Test Actual')
plt.plot(test_df['Date'], test_pred, label='LGBM Forecast')
plt.legend()
plt.grid(True)
plt.show()

# Recursive forecast 45 days
last_known = df.copy()
future_dates = pd.date_range(start=df['Date'].iloc[-1] + pd.Timedelta(days=1), periods=45)
future_pred = []

for date in future_dates:
    temp_df = pd.concat([last_known, pd.DataFrame({'Date': [date], 'Value': [0]})])
    X_future = create_features(temp_df).iloc[-1:]
    pred = model.predict(X_future)[0]
    future_pred.append(pred)
    last_known = pd.concat([last_known, pd.DataFrame({'Date': [date], 'Value': [pred]})])

print("\nLightGBM Forecast for next 45 days:")
for date, val in zip(future_dates, future_pred):
    print(f"{date.date()}: {val:.2f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# UN ORIGINAL DATA - 15 ROWS
data = {
    'Date': ['2024-08-01','2024-08-01','2024-08-01','2024-08-02','2024-08-02',
             '2024-08-02','2024-08-03','2024-08-03','2024-08-03','2024-08-04',
             '2024-08-04','2024-08-04','2024-08-05','2024-08-05','2024-08-05'],
    'Product': ['Laptop','Mouse','Keyboard','Laptop','Mouse','Keyboard',
                'Laptop','Mouse','Keyboard','Laptop','Mouse','Keyboard',
                'Laptop','Mouse','Keyboard'],
    'Quantity': [2,10,5,1,5,3,1,8,2,1,12,4,2,15,6],
    'Price': [50000,500,1500,50000,500,1500,50000,500,1500,50000,500,1500,50000,500,1500]
}

df = pd.DataFrame(data)
df['Total'] = df['Quantity'] * df['Price']
df['Date'] = pd.to_datetime(df['Date'])
df['Day_Number'] = df['Date'].dt.day

# ML Model train pannu
X = df[['Day_Number']]
y = df['Total']
model = LinearRegression()
model.fit(X, y)

print("✅ df and model ready! Ipo dynamic dashboard code run pannu")
print(df.head())
```

```python
!pip install plotly -q

import plotly.graph_objects as go
from plotly.subplots import make_subplots

product_total = df.groupby('Product')['Total'].sum().reset_index()
date_total = df.groupby('Date')['Total'].sum().reset_index()

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Product Wise Total Sales', 'Date Wise Sales Trend',
                    'Price Distribution', 'Sales Prediction'),
    specs=[[{"type": "bar"}, {"type": "scatter"}],
           [{"type": "histogram"}, {"type": "scatter"}]]
)

# Chart 1: Bar - Product Sales
fig.add_trace(go.Bar(
    x=product_total['Product'],
    y=product_total['Total'],
    name='Product Sales',
    marker_color='#FF5733',
    hovertemplate='Product: %{x}<br>Sales: ₹%{y:,.0f}<extra></extra>'
), row=1, col=1)

# Chart 2: Line - Date Trend
fig.add_trace(go.Scatter(
    x=date_total['Date'],
    y=date_total['Total'],
    mode='lines+markers',
    name='Daily Trend',
    line=dict(color='#3498DB', width=3),
    marker=dict(size=10),
    hovertemplate='Date: %{x}<br>Sales: ₹%{y:,.0f}<extra></extra>'
), row=1, col=2)

# Chart 3: Histogram - Price Distribution
fig.add_trace(go.Histogram(
    x=df['Price'],
    name='Price Distribution',
    marker_color='#2ECC71',
    nbinsx=10
), row=2, col=1)

# Chart 4: Scatter + Prediction Line
fig.add_trace(go.Scatter(
    x=df['Day_Number'],
    y=df['Total'],
    mode='markers',
    name='Actual Sales',
    marker=dict(color='blue', size=12)
), row=2, col=2)

# Prediction line
future_days = pd.DataFrame({'Day_Number': [1,2,3,4,5,6]})
future_pred = model.predict(future_days)
fig.add_trace(go.Scatter(
    x=future_days['Day_Number'],
    y=future_pred,
    mode='lines',
    name='Predicted Trend',
    line=dict(color='red', dash='dash', width=3)
), row=2, col=2)

fig.update_layout(
    height=800,
    width=1200,
    title_text="📊 Sales Interactive Dashboard - Dynamic View",
    title_font_size=24,
    showlegend=True,
    hovermode='closest'
)

fig.show()

# HTML ah save pannu
fig.write_html('dynamic_dashboard.html')
print("✅ dynamic_dashboard.html saved! Files panel la irunthu download pannu")
```
